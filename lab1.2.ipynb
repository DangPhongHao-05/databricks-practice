```python
%sql
CREATE SCHEMA IF NOT EXISTS default;

CREATE OR REPLACE TABLE default.orders_delta (
    order_id INT,
    customer_id STRING,
    amount DOUBLE,
    status STRING
)
USING DELTA;
```

```python
%sql
INSERT INTO default.orders_delta VALUES
(1, 'C001', 500000, 'pending'),
(2, 'C002', 700000, 'completed'),
(3, 'C003', 300000, 'pending');
```

```python
%sql
SELECT * FROM default.orders_delta;
```

```python
%sql
-- update
update default.orders_delta set status = 'completed' where order_id = 1;
select * from default.orders_delta;
```

```python
%sql

CREATE TABLE IF NOT EXISTS default.orders_new (
    order_id INT,
    customer_id STRING,
    amount DOUBLE,
    status STRING
)
USING DELTA;
```

```python
%sql

INSERT INTO default.orders_new
VALUES
    (2, 'C002', 1200000, 'completed'),
    (5, 'C005', 650000, 'pending'),
    (6, 'C006', 850000, 'completed');
```

```python
%sql
select * from default.orders_new
```

```python
%sql
--merge oder vs oder_new
merge into default.orders_delta as t
using default.orders_new as s
on t.order_id = s.order_id
when matched then
    update set 
        t.amount = s.amount,
        t.status = s.status
when not matched 
    then insert
        (order_id, customer_id, amount, status)
        values
        (s.order_id, s.customer_id, s.amount, s.status);
    
select * from default.orders_delta order by order_id asc;
    
```

```python
%sql
CREATE TABLE default.orders_clustered (
    order_id INT,
    customer_id STRING,
    amount DOUBLE,
    status STRING
)
USING DELTA
CLUSTER BY (customer_id);
```
